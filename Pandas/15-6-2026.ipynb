```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('deliveries.csv')
data
```

```python
death_over=data[data['over']>15]
death_over
```

```python
all_batsman = death_over.groupby('batsman')['batsman_runs'].count()
x=all_batsman>200
batsman_list=all_batsman[x].index.tolist()
```

```python
final = death_over[death_over['batsman'].isin(batsman_list)]
final.shape
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False)
runs
```

```python
balls=final.groupby('batsman')['ball'].count().sort_values(ascending=False)
balls
```

```python
sr=(runs/balls)*100
sr.sort_values(ascending=False)
```
